```python
#1,2,3
import pandas as pd
import numpy as nm

filepath= "https://raw.githubusercontent.com/laxmimerit/All-CSV-ML-Data-Files-Download/master/IMDB-Movie-Data.csv"
df=pd.read_csv(filepath)
print("\nDataset\n",df)

#Understanding the loaded data
print("\nShape: ",df.shape)
print("\nHead:\n",df.head())
print("\ncolumn names:\n",df.columns)
print("\nDatatype:\n ",df.dtypes)
print("\nMissing values:\n",df.isnull())
```

```python
#4
print("\nNumber of Duplicates:", df.duplicated().sum()) # Check duplicate rows
df = df.drop_duplicates() # remove duplicate
```

```python
#5
# Rename column
df.rename(columns={"Title": "Movie_Title"}, inplace=True)
print("\nUpdated Column Names:")
df.columns
```

```python
#6
# Extract individual genres
unique_genres = df["Genre"].str.split(",").explode().unique()
print("Unique Genres:")
print(unique_genres)


#nuumber of unique genre combination.
print("\nUnique Genre combinations:\n",df["Genre"].unique())
```

```python
#7
df["Genre_Count"] = df["Genre"].str.split(",").str.len()
print(df[["Movie_Title", "Genre_Count"]])
```

```python
#8
print(df[["Revenue (Millions)", "Metascore"]].isnull().sum())

# Fill missing values with mean
df["Revenue (Millions)"] = df["Revenue (Millions)"].fillna(df["Revenue (Millions)"].mean())

df["Metascore"] = df["Metascore"].fillna(df["Metascore"].mean())
print(df[["Revenue (Millions)", "Metascore"]].isnull().sum())
```

```python
#9
top5 = df.sort_values(by="Rating", ascending=False).head(5)

print(top5[["Movie_Title", "Rating"]])
```

```python
#10
print("Movie with Highest Revenue:")
print(df.loc[df["Revenue (Millions)"].idxmax(),["Movie_Title"]])
print("\n\nMovie with Lowest Revenue:")
print(df.loc[df["Revenue (Millions)"].idxmin(),["Movie_Title"]])
```

```python
#11
avg_rating= df.groupby("Director")["Rating"].mean()
avg_rating
```

```python
#12
year=df["Year"].value_counts().sort_index()
print("Number of movies per year: ")
year
```

```python
#13
result=df[(df['Rating']>8)&(df['Revenue (Millions)']>100)]
print(result[["Movie_Title", "Rating", "Revenue (Millions)"]])
```

```python
#14
director_group = df.groupby("Director")[["Rating", "Revenue (Millions)"]].mean()
director_group
```

```python
year_group = df.groupby("Year").agg(Movie_Count=('Year', 'size'), Average_Rating=('Rating', 'mean'))
year_group
```

```python
#16a
df["Profitability"] = df["Revenue (Millions)"] / df["Runtime (Minutes)"]
print(df[["Movie_Title", "Revenue (Millions)", "Runtime (Minutes)", "Profitability"]].head())

#16b
top10 = df.sort_values(by="Profitability", ascending=False).head(10)
print(top10[["Movie_Title", "Profitability"]])
```

```python
#17
conditions = [
    df["Rating"] < 6,
    (df["Rating"] >= 6) & (df["Rating"] < 7.5),
    df["Rating"] >= 7.5
]

categories = ["Poor", "Average", "Excellent"]
df["Rating_Category"] = nm.select(conditions, categories, default='Not Classified')
print(df[["Movie_Title", "Rating", "Rating_Category"]].head(10))
```

```python
#18
def movie_score(row):
    return row["Rating"] * 10 + row["Revenue (Millions)"]
df["Score"] = df.apply(movie_score, axis=1)

print(df[["Movie_Title", "Rating", "Revenue (Millions)", "Score"]].head(10))
```
